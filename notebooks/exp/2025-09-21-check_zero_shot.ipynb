```python
def subset_adata(adata, adata_ref):
    import numpy as np
    import pandas as pd

    # mask presence in ref
    mask = np.isin(adata.obs.ids, adata_ref.obs.ids)
    adata_subset = adata[mask]

    # re-order to match
    desired = adata_ref.obs["ids"].tolist()
    cat = pd.Categorical(adata_subset.obs["ids"], categories=desired, ordered=True)
    adata_subset = adata_subset[adata_subset.obs.assign(_k=cat).sort_values("_k").index, :].copy()

    assert (adata_subset.obs["ids"].values == adata_ref.obs["ids"].values).all()

    return adata_subset

a1 = subset_adata(adata_orig, adata_train)


```

```python
import numpy as np
from anndata import AnnData


def _digitize_with_tiebreak(x: np.ndarray, bins: np.ndarray) -> np.ndarray:
    """
    Digitize with random tie-breaking when bins collapse (like scGPT).
    """
    left_digits = np.digitize(x, bins)
    right_digits = np.digitize(x, bins, right=True)

    # if bins are identical, spread assignments randomly
    rands = np.random.rand(len(x))
    digits = rands * (right_digits - left_digits) + left_digits
    return np.ceil(digits).astype(np.int64)


def independent_binning_scGPT(
    adata: AnnData,
    layer: str = "X",
    n_bins: int = 51,
    result_key: str = "X_binned",
    save_edges: bool = True,
) -> AnnData:
    """
    Reimplementation of scGPT's binning logic:
    - zeros always bin 0
    - non-zeros quantile-binned per row
    - stochastic tie-breaking for flat quantiles
    """
    data = adata.layers[layer] if layer in adata.layers else adata.X
    data = data.toarray() if hasattr(data, "toarray") else np.array(data)

    binned_rows = []
    bin_edges = []

    for row in data:
        if row.max() == 0:  # all zeros
            binned_rows.append(np.zeros_like(row, dtype=np.int64))
            bin_edges.append(np.zeros(n_bins))
            continue

        non_zero_ids = np.nonzero(row)[0]
        non_zero_vals = row[non_zero_ids]

        bins = np.quantile(non_zero_vals, np.linspace(0, 1, n_bins - 1))

        non_zero_digits = _digitize_with_tiebreak(non_zero_vals, bins)

        binned_row = np.zeros_like(row, dtype=np.int64)
        binned_row[non_zero_ids] = non_zero_digits
        binned_rows.append(binned_row)

        bin_edges.append(np.concatenate([[0], bins]))

    adata.layers[result_key] = np.stack(binned_rows)
    if save_edges:
        adata.obsm[f"{result_key}_edges"] = np.stack(bin_edges)

    return adata
```
